F1 has had many different points systems throughout the years. The current point system (25 points for 1st cascading down to 1 point for 10th) started in 2010. While there have been minor changes to the points system since, the distribution of points for race finishes has remained the same. This will be treated as the modern era for F1.

From 1950 to 1959 F1 awarded points to the top 5 drivers.

From 1961 to 2002 F1 awarded points to the top 6 drivers.

From 2003 to 2009 F1 awarded points to the top 8 drivers.

While there were minor changes in between as periods, the distribution of points for race finishes has remained the same.

A player's elo only changes when they win or lose, gaining and losing elo respectively. While the first place driver is the sole winner, elo will be awarded to all points winning drivers. Their elo gain will be adjusted for their finishing postion. This applies to drivers outside of the points as well. For exmaple, a driver that barely missed out on points will lose less elo than one who finished near the end of the grid.
